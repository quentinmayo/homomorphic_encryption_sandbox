# Salary Experiment (BFV) Walkthrough

This notebook encrypts employee salary and company contribution data, computes total compensation under BFV, and compares against plaintext baselines.

```python
from pathlib import Path
import sys

repo_root = Path.cwd()
if not (repo_root / "src").exists():
    repo_root = repo_root.parent

src_path = str(repo_root / "src")
if src_path not in sys.path:
    sys.path.insert(0, src_path)
```

```python
import importlib

from he_sandbox.experiments.salary.scenario import sample_employees
import he_sandbox.experiments.salary.bfv_salary as bfv_salary

importlib.reload(bfv_salary)

plaintext_totals = bfv_salary.plaintext_totals
encrypt_totals = bfv_salary.encrypt_totals
decrypt_totals = bfv_salary.decrypt_totals
```

```python
employees = sample_employees()
for employee in employees:
    print(employee)
```

```python
plain_per_employee, plain_total = plaintext_totals(employees)
enc_per_employee_cipher, enc_total_cipher = encrypt_totals(employees)

print("Plain per-employee totals:", plain_per_employee)
print("Cipher object type:", type(enc_per_employee_cipher).__name__)
print("Cipher object repr (truncated):", str(enc_per_employee_cipher)[:120] + "...")
```

```python
cipher_blob = enc_per_employee_cipher.serialize()
print("Serialized ciphertext size (bytes):", len(cipher_blob))
print("First 32 bytes:", cipher_blob[:32])

enc_per_employee, enc_total = decrypt_totals(enc_per_employee_cipher, enc_total_cipher)
print("Decrypted per-employee totals:", enc_per_employee)
print("Decrypted company total:", enc_total)
```

```python
assert enc_per_employee == plain_per_employee
assert enc_total == plain_total
print("BFV totals match plaintext exactly.")
```
