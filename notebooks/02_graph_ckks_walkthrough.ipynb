# Graph Experiment (CKKS) Walkthrough

This notebook encrypts graph node weights, computes one-step weighted neighbor scores, and compares encrypted outputs against plaintext results with tolerance checks.

```python
from pathlib import Path
import sys

repo_root = Path.cwd().parent
sys.path.append(str(repo_root / "src"))
```

```python
from he_sandbox.experiments.graph.ckks_graph import plaintext_neighbor_score, encrypted_neighbor_score
```

```python
adjacency = [
    [0.0, 0.8, 0.2],
    [0.5, 0.0, 0.5],
    [0.2, 0.6, 0.0],
]
node_weights = [2.0, 1.0, 3.0]
```

```python
plain_scores = plaintext_neighbor_score(adjacency, node_weights)
enc_scores = encrypted_neighbor_score(adjacency, node_weights)

print("Plain scores:", plain_scores)
print("Encrypted scores:", enc_scores)
```

```python
tolerance = 1e-3
for i, (enc_value, plain_value) in enumerate(zip(enc_scores, plain_scores, strict=True)):
    diff = abs(enc_value - plain_value)
    print(f"Node {i}: diff={diff:.8f}")
    assert diff <= tolerance
print("CKKS outputs are within tolerance.")
```
